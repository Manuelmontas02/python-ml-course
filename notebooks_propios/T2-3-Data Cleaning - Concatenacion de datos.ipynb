## Concatenar cientos de datos distribuidos

```python
data = pd.read_csv("..\datasets\distributed-data\\001.csv")
data.head()
```

```python
data.shape
```

* Importar el primer fichero
* Hacemos un buble para ir recorriendo todos y cada uno de los ficheros. Importante hay que tener una consistencia en el nombre de los fichero.
* Cada uno de ellos debe apendizarse del primer fichero que ya habiamos cargado
* Repetimos el bucle hasta que no queden ficheros



```python
filepath = r"..\datasets\distributed-data\\"
data = pd.read_csv("..\datasets\distributed-data\001.csv")

for i in range(2,333):
    if i < 10:
        filename = "00" + str(i)
    if 10 <= i < 100:
        filename = "0" + str(i)
    if i >= 100:
        filename = str(i)
    file = filepath + filename + ".csv"
    temp_data = pd.read_csv(file)

    data = pd.concat([data,temp_data],axis = 0)
```
