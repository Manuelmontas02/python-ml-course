# Agregacion de datos por categoria

```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
```

```python
gender = ["Male","Female"]
income = ["Poor","Midle Class","Rich"]
```

```python
n = 500
gender_data = []
income_data = []

for i in range(0,n):
    gender_data.append(np.random.choice(gender))
    income_data.append(np.random.choice(income))
income_data[0:10]
```

```python
height = 160 + 30 * np.random.randn(n)
weight = 65 + 25 * np.random.randn(n)
age = np.ceil(30 + 12 * np.random.randn(n)).astype(int)
income = 18000 + 3500*np.random.randn(n)
```

```python
data = pd.DataFrame(
    {
        'Gender': gender_data,
        'Economic Status': income_data,
        'Height': height,
        'Weight': weight,
        'Age': age,
        'Income': income
    }
)
data.head()
```

### Agrupacion de datos

```python
grouped_gender = data.groupby("Gender")
```

```python
grouped_gender.groups
```

```python
for names, groups in grouped_gender:
    print(names)
    print(groups)
```

```python
grouped_gender.get_group("Female")
```

```python
double_group = data.groupby(["Gender","Economic Status"])
double_group
```

```python
len(double_group)
```

```python
for names, groups in double_group:
    print(names)
    print(groups)
```

## Operaciones de agregacion sobre datos agrupados

```python
double_group.sum()
```

```python
double_group.mean()
```

```python
double_group.size()
```

```python
double_group.describe()
```

```python
double_group["Income"].describe()
```

```python
double_group.aggregate(
    {
    "Income":"sum",
    "Age":"mean",
    "Height":"std"
    }
)
```

Actualmente se utiliza dentro de la funcion de aggregate de esta manera. De acuerdo al tutorial, seria por ejemplo, "Income":np.sum, pero por modificaciones en python se esta utilizando "Income":"sum"

```python
double_group.aggregate(
    {
        "Age": "mean",
        "Height":lambda h:np.mean(h)/np.std(h) # La definicion de un lambda se da cuando quiero ejecutar una operacion personalizada
    }
)
```

```python
double_group.aggregate(["sum","mean","std"])
```

```python
double_group.aggregate([lambda x: np.mean(x)/np.std(x)])
```

### Descripcion de metricos para los grupos "Female" y "Middle Class"

```python
double_group.get_group(("Female","Midle Class")).describe()
```

## Filtrado de datos

```python
double_group["Age"].filter(lambda x: x.sum()>2400)
```

## Transformacion de variables

```python
zscore = lambda x: (x - x.mean())/x.std()
```

```python
z_group = double_group.transform(zscore)
```

```python
plt.hist(z_group["Age"])
plt.xlabel("Distribution")
plt.ylabel("Frequency")
plt.title("Distribution")
```

#### La funcion de transform puede ser usada para el rellenado de Nas

```python
## Ejemplo

fill_na_mean = lambda x: x.fillna(x.mean())
double_group.transform(fill_na_mean)
```
