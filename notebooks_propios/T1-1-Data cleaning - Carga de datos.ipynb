### Descargar y procesar datos desde una URL externa (Completo)

```python
# Obtencion de la direccion URL
olympics_url = "https://raw.githubusercontent.com/brandomr/olympics/refs/heads/master/winterolympics.csv"
# Ejercicio completo
import urllib3
http = urllib3.PoolManager()
r = http.request('GET',olympics_url)
print("El status de la respuesta es: %d" % (r.status))
response = r.data

# El objeto response contiene un string binario, asi que lo convertimos a un string descodificandolo en UTF-8
str_data = response.decode('utf-8')

# Dividimos el string en un array de filas, separandolo por intros
lines = str_data.split("\n")

# La primera linea contiene la cabecera asi que la estraemos
col_names = lines[0].split(",")

#Generamos un diccionario vacio donde ira la informacion procesada desde la URL externa
counter = 0
main_dict = {}
for col in col_names:
    main_dict[col] = []

# Procesamos fila a fila la informacion para ir rellenando el diccionario con los datos como hicimos antes
for line in lines:
    # Nos saltamos la primera linea que es la que contiene la cabecera
    if(counter > 0):
        # Separamos por comas y limpiamos espacios en blanco 
        values = line.strip().split(",")
        # Se anade cada valor a su respectiva columna del diccionario
        for i in range(len(col_names)):
            main_dict[col_names[i]].append(values[i])
    counter += 1

print("El numero de filas %d y el numero de columnas %d " % (counter,len(col_names)))

# Convertir de un dataframe
medals_df = pd.DataFrame(main_dict)

# seleccionar donde guardar el dataframe creado 
mainpath = r"C:\Users\USER\OneDrive\Documentos\GitHub\python-ml-course\\"
filepath = r"notebooks_propios\Datasets_generados\medals_df"
fullpath = os.path.join(mainpath,filepath)

# Lo guardamos en diferentes formatos
medals_df.to_csv(fullpath+".csv")
medals_df.to_json(fullpath+".json")
medals_df.to_excel(fullpath+".xlsx")
```

## Crear una funcion para hacer esto generico

```python
def downloadFromURL(url,filename,sep = ",",delim = "\n",encoding = 'utf-8'):
    # Obtencion de la direccion URL
    olympics_url = "https://raw.githubusercontent.com/brandomr/olympics/refs/heads/master/winterolympics.csv"
    # Ejercicio completo
    import urllib3
    http = urllib3.PoolManager()
    r = http.request('GET',url)
    print("El status de la respuesta es: %d" % (r.status))
    response = r.data
    
    # El objeto response contiene un string binario, asi que lo convertimos a un string descodificandolo en UTF-8
    str_data = response.decode('utf-8')
    
    # Dividimos el string en un array de filas, separandolo por intros
    lines = str_data.split(delim)
    
    # La primera linea contiene la cabecera asi que la estraemos
    col_names = lines[0].split(sep)
    
    #Generamos un diccionario vacio donde ira la informacion procesada desde la URL externa
    counter = 0
    main_dict = {}
    for col in col_names:
        main_dict[col] = []
    
    # Procesamos fila a fila la informacion para ir rellenando el diccionario con los datos como hicimos antes
    for line in lines:
        # Nos saltamos la primera linea que es la que contiene la cabecera
        if(counter > 0):
            # Separamos por comas y limpiamos espacios en blanco 
            values = line.strip().split(sep)
            # Se anade cada valor a su respectiva columna del diccionario
            for i in range(len(col_names)):
                main_dict[col_names[i]].append(values[i])
        counter += 1
    
    print("El numero de filas %d y el numero de columnas %d " % (counter,len(col_names)))
    
    # Convertir de un dataframe
    df = pd.DataFrame(main_dict)
    
    # seleccionar donde guardar el dataframe creado 
    mainpath = r"C:\Users\USER\OneDrive\Documentos\GitHub\python-ml-course\notebooks_propios\Datasets_generados\\"
    #filename = r"notebooks_propios\Datasets_generados\\"
    fullpath = os.path.join(mainpath,filename)
    
    # Lo guardamos en diferentes formatos
    df.to_csv(fullpath+".csv")
    df.to_json(fullpath+".json")
    df.to_excel(fullpath+".xlsx")
    return df
```
