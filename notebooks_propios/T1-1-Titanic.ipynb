## Carga de datos a través de la función open

Open es muy bueno cuando el dataset es muy grande

```python
data3 = open(mainpath + "\\" + r"customer-churn-model\Customer Churn Model.txt","r")
```

```python
cols = next(data3).strip().split(",")
n_cols = len(cols)
```

```python
counter = 0
main_dict = {}
for col in cols:
    main_dict[col] = []
```

```python
counter = 0
for line in data3:
    values = line.strip().split(",")
    for i in range(len(cols)):
        main_dict[cols[i]].append(values[i])
    counter += 1

print ("El dataset tiene %d filas y %d columnas" % (counter,n_cols))   
```

## Explicacion del codigo

Tomo los valores de la primera fila, elimino los espacios en blancos y caracteres especiales  usando el metodo .strip(), luego separo por "," usando split(","). 

Recorro columna a columna llenando cada uno de los espacios vacios del directorio para esa fila, luego paso a la siguiente linea.

```python
df3 = pd.DataFrame(main_dict)
df3.head()
```

## Escritura


```python
infile = mainpath + "\\" + r"customer-churn-model\Customer Churn Model.txt"
outfile = mainpath + "\\" + r"customer-churn-model\Tab Customer Churn Model.txt"
```

```python
with open(infile,"r") as infile1:
    with open(outfile,"w") as outfile1:
        for line in infile1:
            files = line.strip().split(",")
            outfile1.write("\t".join(files))
```
