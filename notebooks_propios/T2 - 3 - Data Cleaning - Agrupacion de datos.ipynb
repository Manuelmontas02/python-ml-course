# Agregacion de datos por categoria

```python
import numpy as np
import pandas as pd
```

```python
gender = ["Male","Female"]
income = ["Poor","Midle Class","Rich"]
```

```python
n = 500
gender_data = []
income_data = []

for i in range(0,n):
    gender_data.append(np.random.choice(gender))
    income_data.append(np.random.choice(income))
income_data[0:10]
```

```python
height = 160 + 30 * np.random.randn(n)
weight = 65 + 25 * np.random.randn(n)
age = np.ceil(30 + 12 * np.random.randn(n)).astype(int)
income = 18000 + 3500*np.random.randn(n)
```

```python
data = pd.DataFrame(
    {
        'Gender': gender_data,
        'Economic Status': income_data,
        'Height': height,
        'Weight': weight,
        'Age': age,
        'Income': income
    }
)
data.head()
```
